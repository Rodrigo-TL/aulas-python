```python
import pandas as pd
import numpy as np

# 1. Criando a base de dados
dados_alunos = [
    {"nome": "João", "turno": "N", "nota": 8.5},
    {"nome": "Marcelo", "turno": "T", "nota": 7.5},
    {"nome": "Victor", "turno": "N", "nota": 5.5},
    {"nome": "Bruno", "turno": "M", "nota": 9.5},
    {"nome": "Rafael", "turno": "M", "nota": 10.0}
]

df_alunos = pd.DataFrame(dados_alunos)

# 2. Inserindo coluna por mapeamento de turno
map_turnos = {
    "M": "Manhã",
    "T": "Tarde",
    "N": "Noite"
}
df_alunos["turno_completo"] = df_alunos["turno"].map(map_turnos)

# 3. Criando a coluna de status (criamos uma nova para não perder as notas numéricas)
df_alunos["status"] = np.where(
    df_alunos["nota"] >= 5.0, "Aprovado", "Reprovado"
)

# 4. Criando a coluna de troca de turno com base no status correto
df_alunos["troca_turno"] = np.where(
    df_alunos["status"] == "Aprovado",
    "Concluiu o ensino médio",
    "Tem que fazer a prova de recuperação"
)

# 5. Salvando o arquivo CSV
df_alunos.to_csv("alunos_data.csv", index=False)

# Exibindo o resultado final
print(df_alunos)
```
