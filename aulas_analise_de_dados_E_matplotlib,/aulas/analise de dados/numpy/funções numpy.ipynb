```python
import numpy as np
import pandas as pd

# 1. Operações Básicas com Renda e Gastos
lista_renda = [5000, 4500, 6000, 4000]
lista_gastos = [4000, 3500, 6000, 3900]

renda = np.array(lista_renda)
gastos = np.array(lista_gastos)

lucro = renda - gastos
print(f"Lucro dos meses: {lucro}")

# 2. Filtragem de Notas de Alunos (Máscaras Booleanas)
notas_alunos = np.array([10.0, 6.0, 5.6, 7.4, 8.8])

notas_aprovados = notas_alunos[notas_alunos >= 7]
print(f"Notas aprovadas: {notas_aprovados}")

notas_reprovados = notas_alunos[~(notas_alunos >= 7)]
print(f"Notas reprovadas: {notas_reprovados}")

# 3. Operações em Matrizes Bidimensionais (Vendas por Trimestre)
t1 = [15000, 20000, 30000]
t2 = [25000, 30000, 40000]
t3 = [35000, 40000, 50000]
t4 = [45000, 50000, 60000]

vendas = np.array([t1, t2, t3, t4])

total_vendas_ano = np.sum(vendas)
print(f"Total de vendas no ano: {total_vendas_ano}")

# axis=1 -> soma por linha (total de cada trimestre)
total_por_trimestre = np.sum(vendas, axis=1)
print(f"Total por trimestre: {total_por_trimestre}")

# axis=0 -> soma por coluna
comeco_de_trimestre = np.sum(vendas, axis=0)
print(f"Soma por meses do bloco: {comeco_de_trimestre}")

maior_venda_trimestre = np.max(vendas, axis=1)
print(f"Maior venda por trimestre: {maior_venda_trimestre}")

menor_venda_ano = np.min(vendas)
print(f"Menor venda ano: {menor_venda_ano}")

# 4. Condicionais com np.where em Arrays de Valores
lista_valores = [10, 15, 20, 30]
valor_produtos = np.array(lista_valores)

valor_promocao = np.where(
    valor_produtos >= 20, valor_produtos * 0.9, valor_produtos
)
valor_promocao = np.round(valor_promocao, 2)
print(f"Valores promocionais: {valor_promocao}")

# 5. Manipulação Avançada de DataFrames (Pandas + NumPy)
estoque_inicial = [10, 0, 5, 9]
valores_iniciais = [2000, 1500, 500, 5000]

df_produtos = pd.DataFrame({
    "Produto": ['Notebook', 'Celular', 'Monitor', "Notebook"],
    "Estoque": estoque_inicial,
    "Valor": valores_iniciais,
    "Cliente": ["Magalu", "Americanas", "Kabum", "Magalu"]
})

# Desconto de 10% para valores acima de 1500
df_produtos["Valor"] = np.where(
    df_produtos["Valor"] > 1500,
    df_produtos["Valor"] * 0.9,
    df_produtos["Valor"]
)

# Desconto extra de 20% para o cliente Kabum
df_produtos["Valor"] = np.where(
    df_produtos["Cliente"] == "Kabum",
    df_produtos["Valor"] * 0.8,
    df_produtos["Valor"]
)

# Regra de atualização do Estoque baseada no Cliente
df_produtos["Estoque"] = np.where(
    df_produtos["Cliente"] == "Kabum",
    df_produtos["Estoque"] * 0,
    np.where(
        df_produtos["Cliente"] == "Magalu",
        df_produtos["Estoque"] - 2,
        df_produtos["Estoque"]
    )
)

# Definição do status de disponibilidade
df_produtos["Status"] = np.where(
    df_produtos["Estoque"] > 0,
    "Disponível",
    "Esgotado"
)

print("\nDataFrame final de produtos:")
print(df_produtos)

# Agrupando e somando os valores por Cliente e Estoque
agrupamento_count = df_produtos.groupby(["Cliente", "Estoque"])["Valor"].sum()
print("\nSoma dos valores agrupados por Cliente e Estoque:")
print(agrupamento_count)
```
